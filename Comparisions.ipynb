```python
import matplotlib.pyplot as plt
import numpy as np

# Data
your_model_languages = ['Hindi', 'Spanish', 'English']
your_model_accuracy = [49.36102236421725, 58.9585172109444,  57.89233576642335]

other_model_languages = ['Arabic', 'chinese','French']
other_model_accuracy = [55, 50, 56,]

# Combine languages for a grouped bar plot
languages = your_model_languages + other_model_languages
x = np.arange(len(languages))

# Create bar chart
plt.figure(figsize=(10, 6))
bar_width = 0.4

plt.bar(x[:3], your_model_accuracy, width=bar_width, label='Language specific', color='skyblue')
plt.bar(x[3:], other_model_accuracy, width=bar_width, label='Twitter-Roberta', color='orange')

# Customizations
plt.title('Accuracy Comparison Between Models', fontsize=14)
plt.ylabel('Accuracy (%)', fontsize=12)
plt.xticks(x, languages, rotation=45, fontsize=10)
plt.ylim(0, 65)
plt.legend(fontsize=10)
plt.tight_layout()

# Show the plot
plt.show()
```
