```python
import re

# Function to clean text
def clean_text(text):
    # Convert text to lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(r'http\S+|www\S+|https\S+', '', text, flags=re.MULTILINE)

    # Remove email addresses
    text = re.sub(r'\S*@\S*\s?', '', text)

    # Remove mentions (@usernames)
    text = re.sub(r'@\w+', '', text)

    # Remove hashtags
    text = re.sub(r'#\w+', '', text)

    # Remove special characters and numbers, keep only letters
    text = re.sub(r'[^a-zA-Z\s]', '', text)

    # Remove extra whitespaces
    text = re.sub(r'\s+', ' ', text).strip()

    return text

# Apply the cleaning function to the DataFrame
df_tw_train['text'] = df_tw_train['text'].apply(clean_text)

# View the cleaned DataFrame
df_tw_train.head()
```

```python
#Removoing stopwords
def remove_stopwords(text, language):
  stop_words = set(stopwords.words(language))
  word_tokens = word_tokenize(text)
  filtered_sentence = [w for w in word_tokens if not w.lower() in stop_words]
  return " ".join(filtered_sentence)


df_tw_train['text'] = df_tw_train.apply(lambda row:
                               remove_stopwords(row['text'], 'english') if row['text'] is not None else '', axis=1)
df_tw_train.head()
```

```python
!pip install textblob
!python -m textblob.download_corpora

from textblob import Word
from textblob import TextBlob
import pandas as pd


# Function to lemmatize text using TextBlob
def lemmatize_text(text):
    # Break text into words, lemmatize each word, and join back into a sentence
    blob = TextBlob(text)
    return ' '.join([Word(word).lemmatize() for word in blob.words])

# Apply the lemmatization function to the 'text' column
df_tw_train['text'] = df_tw_train['text'].apply(lemmatize_text)

# Print the updated DataFrame
df_tw_train
```
